```r
library(readr) # library for function read Data

mydata <- read_csv("sample_data/makkah_data.csv") # Call Data
#print(mydata)

data_of_makkah <- subset(mydata, Region == "Makkah") # Value Allocation


library(dplyr) # library for function arrange Data

data_of_makkah <- data_of_makkah %>% arrange(desc(Year)) # arrange data

print(data_of_makkah, n = nrow(data_of_makkah)) # function for print all of rows
write.csv(data_of_makkah, file = "makkah_data.csv", row.names = FALSE) # save the Work
```

```r
# Load the tidyverse library for visualization and data manipulation
library(tidyverse)

# Create a data frame with the population data and years
data <- data.frame(
  Year = 2010:2022,
  Saudi_Population = c(
    3077186, 3145642, 3619353, 3308457, 3174883, 3505532,
    3595076, 3675514, 3763757, 3858847, 3964569, 4059151, 4153723
  ),
  Non_Saudi_Population = c(
    3272031, 2409610, 2392803, 3982721, 4993437, 4298604,
    4998534, 4344243, 3951581, 3493709, 4017960, 3624037, 3868740
  )
)

# Reshape the data for plotting
data <- data %>%
  pivot_longer(cols = -Year, names_to = "Population_Type", values_to = "Population")

# Create a bar chart with years on the Y-axis and population on the X-axis
ggplot(data, aes(x = Year, y = Population, fill = Population_Type)) +
  geom_bar(stat = "identity", position = "dodge", width = 0.8) +
  scale_fill_manual(values = c("Saudi_Population" = "gray", "Non_Saudi_Population" = "red")) +
  labs(title = "Population Distribution by Year for Problem 1",
       x = "Year",
       y = "Population") +
  theme_minimal()











```

```r
# Create a data frame with the population data and years for Saudi Men and Non-Saudi Men
data <- data.frame(
  Year = 2010:2022,
   Saudi_Females = c(
    1535534, 1570069, 1608752, 1653242, 1703946, 1753158, 1798091, 1838533, 1882931, 1931101, 1984600, 2032042, 2079845
  ),
  Non_Saudi_Females = c(
    1076909, 1094964, 1107492, 1155014, 1163579, 1185309, 1199669, 1196162, 1103527, 1053771, 1153302, 1049612, 1062692
  )
)

# Load the ggplot2 library for visualization
library(ggplot2)

# Create a line chart with years on the X-axis and population on the Y-axis
ggplot(data, aes(x = Year)) +
  geom_line(aes(y = Saudi_Females, color = "Saudi females")) +
  geom_line(aes(y = Non_Saudi_Females, color = "Non-Saudi females")) +
  labs(title = "Population of Saudi and Non-Saudi women Over the Years",
       x = "Year",
       y = "Population") +
  scale_color_manual(values = c("Saudi females" = "blue", "Non-Saudi females" = "red")) +
  theme_minimal()
```
